```python
import spacy
from sklearn.metrics import classification_report, accuracy_score
```

**Load model**

```python
nlp = spacy.load("en_core_web_sm")
```

**Sample Text**

```python
texts = [
    "Tesla opened a new office in Mumbai.",
    "Sundar Pichai met Narendra Modi in Delhi.",
    "Amazon invested $3 million in a startup based in Bangalore.",
    "Cristiano Ronaldo visited Paris last week.",
    "Infosys is collaborating with IBM in New York."
]
```

**True entities**

```python
true_entities = [
    [("Tesla", "ORG"), ("Mumbai", "GPE")],
    [("Sundar Pichai", "PERSON"), ("Narendra Modi", "PERSON"), ("Delhi", "GPE")],
    [("Amazon", "ORG"), ("$3 million", "MONEY"), ("Bangalore", "GPE")],
    [("Cristiano Ronaldo", "PERSON"), ("Paris", "GPE"), ("last week", "DATE")],
    [("Infosys", "ORG"), ("IBM", "ORG"), ("New York", "GPE")]
]
```

```python
y_true = []
y_pred = []

for text, gold_entities in zip(texts, true_entities):
    doc = nlp(text)

    # Convert gold entities to dict
    gold_dict = {ent[0]: ent[1] for ent in gold_entities}

    for ent in doc.ents:
        if ent.text in gold_dict:
            y_true.append(gold_dict[ent.text])
            y_pred.append(ent.label_)

    # Handle missed entities
    for gold_ent, gold_label in gold_dict.items():
        if gold_ent not in [ent.text for ent in doc.ents]:
            y_true.append(gold_label)
            y_pred.append("O")  # Not detected
```

**Print results**

```python
print("Accuracy:", accuracy_score(y_true, y_pred))
print("\nDetailed Classification Report:\n")
print(classification_report(y_true, y_pred))
```
